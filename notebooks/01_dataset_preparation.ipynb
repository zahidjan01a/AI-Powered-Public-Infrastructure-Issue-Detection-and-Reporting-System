```python
import os
import zipfile
```

```python
raw_dir = "../dataset/raw"
processed_dir = "../dataset/processed"

os.makedirs(processed_dir, exist_ok=True)
```

```python
for file in os.listdir(raw_dir):

    if file.endswith(".zip"):

        zip_path = os.path.join(raw_dir, file)

        extract_folder = os.path.join(
            processed_dir,
            file.replace(".zip", "")
        )

        os.makedirs(extract_folder, exist_ok=True)

        with zipfile.ZipFile(zip_path, "r") as zip_ref:
            zip_ref.extractall(extract_folder)

        print(f"Extracted: {file}")
```

```python
import os

processed_dir = "../dataset/processed"

for folder in os.listdir(processed_dir):
    folder_path = os.path.join(processed_dir, folder)

    print("=" * 60)
    print("Dataset:", folder)

    for root, dirs, files in os.walk(folder_path):
        level = root.replace(folder_path, "").count(os.sep)

        indent = "    " * level

        print(f"{indent}{os.path.basename(root)}/")

        # Show only first 2 levels
        if level >= 2:
            break
```

```python
import os

processed_dir = "../dataset/processed"

for dataset in os.listdir(processed_dir):

    dataset_path = os.path.join(processed_dir, dataset)

    print("\n" + "="*60)
    print(f"DATASET: {dataset}")

    for root, dirs, files in os.walk(dataset_path):

        level = root.replace(dataset_path, "").count(os.sep)

        if level <= 3:
            print("    "*level + os.path.basename(root))

        if level == 3:
            break
```

```python
import os

processed_dir = "../dataset/processed"

image_extensions = (".jpg", ".jpeg", ".png", ".bmp")

for dataset in os.listdir(processed_dir):

    dataset_path = os.path.join(processed_dir, dataset)

    total_images = 0

    for root, dirs, files in os.walk(dataset_path):

        images = [
            file for file in files
            if file.lower().endswith(image_extensions)
        ]

        total_images += len(images)

    print(f"{dataset} : {total_images} images")
```
